Etapa 1

```python
with open('actors.csv', mode='r') as file:
    lines = file.readlines()

header = lines[0].strip().split(',')
data = [line.strip().split(',') for line in lines[1:]]

number_of_movies_index = header.index('Number of Movies')
actor_index = header.index('Actor')

max_movies = str('0')
top_actor = ""

for row in data:
    num_movies = str(row[number_of_movies_index])
    if num_movies > max_movies:
        max_movies = num_movies
        top_actor = row[actor_index]

with open('etapa-1.txt', 'w') as file:
        file.write(f'O ator com mais filmes foi: {top_actor}, com: {max_movies} filmes\n')
```

Etapa 2

```python
with open('actors.csv', mode='r') as file:
    lines = file.readlines()

for i, linha in enumerate(lines):
    if i == 5: 
        posicao_primeira_virgula = linha.find(',')

        linha_modificada = linha[:posicao_primeira_virgula] + linha[posicao_primeira_virgula+1:]

        lines[i] = linha_modificada

header = lines[0].strip().split(',')
data = [line.strip().split(',') for line in lines[1:]]

gross_index = header.index('Gross')

total_gross = 0
count = 0

for row in data:
    gross_value = float(row[gross_index])
    total_gross += gross_value
    count += 1

average_gross = total_gross / count

with open('etapa-2.txt', 'w') as file:
    file.write(f'A média da coluna Gross é {average_gross:.2f} milhões de dólares.')
```

Etapa 3

```python
with open('actors.csv', mode='r') as file:
    lines = file.readlines()

for i, linha in enumerate(lines):
    if i == 5:  
        posicao_primeira_virgula = linha.find(',')

        if posicao_primeira_virgula != -1:
            linha_modificada = linha[:posicao_primeira_virgula] + linha[posicao_primeira_virgula+1:]

            lines[i] = linha_modificada

header = lines[0].strip().split(',')
data = [line.strip().split(',') for line in lines[1:]]

average_per_movie_index = header.index('Average per Movie')

maior_media = 0
melhor_ator = None

for row in data:
    average_per_movie = float(row[average_per_movie_index])
    if average_per_movie > maior_media:
        maior_media = average_per_movie
        melhor_ator = row[header.index('Actor')]

with open('etapa-3.txt', 'w') as file:
    file.write(f'O ator/atriz com a maior média de receita bruta por filme é {melhor_ator} com uma média de {maior_media:.2f} milhões de dólares.')
```

Etapa 4

```python
with open('actors.csv', 'r') as arquivo:
    linhas = arquivo.readlines()

    for i, linha in enumerate(linhas):
        if i == 5:
            posicao_primeira_virgula = linha.find(',')

            if posicao_primeira_virgula != -1:
                linha_modificada = linha[:posicao_primeira_virgula] + linha[posicao_primeira_virgula + 1:]

                linhas[i] = linha_modificada

dados_csv = [linha.strip().split(',') for linha in linhas][1:]

contagem_filmes = {}

for linha in dados_csv:
    filme = linha[4]
    if filme in contagem_filmes:
        contagem_filmes[filme] += 1
    else:
        contagem_filmes[filme] = 1

filmes_ordenados = sorted(contagem_filmes.items(), key=lambda x: (-x[1], x[0]))

with open('etapa-4.txt', 'w') as arquivo_saida:
    for i, (filme, quantidade) in enumerate(filmes_ordenados, 1):
        linha = f"{i} - O filme ({filme}) aparece {quantidade} vez(es) no dataset\n"
        arquivo_saida.write(linha)
```

Etapa 5

```python
with open('actors.csv', 'r') as file:
    lines = file.readlines()

for i, linha in enumerate(lines):
    if i == 5:  
        posicao_primeira_virgula = linha.find(',')

        if posicao_primeira_virgula != -1:
            linha_modificada = linha[:posicao_primeira_virgula] + linha[posicao_primeira_virgula+1:]

            lines[i] = linha_modificada

headers = lines[0].strip().split(',')
data = [line.strip().split(',') for line in lines[1:]]

total_gross_index = headers.index('Total Gross')

data.sort(key=lambda x: float(x[total_gross_index]), reverse=True)

with open('etapa-5.txt', 'w') as file:
    for row in data:
        actor_name = row[headers.index('Actor')]
        total_gross = row[total_gross_index]
        file.write(f"{actor_name} - {total_gross}\n")

```
