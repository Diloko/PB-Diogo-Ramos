1 - Removendo linhas dupllicadas utilizando a biblioteca pandas

```python
import pandas as pd

df = pd.read_csv('googleplaystore.csv')

df_sem_duplicatas = df.drop_duplicates()

df_sem_duplicatas.to_csv('googleplay_sem_duplicatas.csv', index=False)
```

2 - Gráfico de barras contendo os top 5 apps por n° instalação

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('googleplay_sem_duplicatas.csv')

df = df[df['Installs'].str.contains('^[0-9,+]+$')]  
df['Installs'] = df['Installs'].str.replace('[+,]', '', regex=True).astype(int)

top5 = df.nlargest(5, 'Installs')[['App', 'Installs']]

plt.figure(figsize=(10, 6))
bars = plt.bar(top5['App'], top5['Installs'], color='skyblue')

for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval, int(yval), va='bottom')  

plt.xlabel('Aplicativos')
plt.ylabel('Instalações')
plt.title('Top 5 Aplicativos Mais Instalados')
plt.xticks(rotation=45)
plt.show()
```

3 - Gráfico de pizza mostrando as categorias de apps no dataset

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('googleplay_sem_duplicatas.csv')

categoria_contagem = df['Category'].value_counts()

plt.figure(figsize=(30, 18))
wedges, texts, autotexts = plt.pie(categoria_contagem, labels=categoria_contagem.index, autopct='%1.1f%%', startangle=140, colors=plt.cm.Paired.colors)

plt.legend(wedges, categoria_contagem.index, title="Categorias", loc="center left", bbox_to_anchor=(1, 0, 0.5, 1))

plt.title('Distribuição das Categorias')
plt.axis('equal')  
plt.show()
```

4 - Aplicativo mais caro no dataset

```python
import pandas as pd

df = pd.read_csv('googleplay_sem_duplicatas.csv')

df['Price'] = pd.to_numeric(df['Price'].str.replace('$', ''), errors='coerce')

df = df.dropna(subset=['Price'])

app_mais_caro = df.loc[df['Price'].idxmax()]

print("O aplicativo mais caro é:")
print(app_mais_caro)
```

6 - Top 10 apps por n° de reviews

```python
import pandas as pd

df = pd.read_csv('googleplay_sem_duplicatas.csv')

df['Reviews'] = pd.to_numeric(df['Reviews'], errors='coerce')

df = df.dropna(subset=['Reviews'])

df = df.sort_values(by='Reviews', ascending=False).drop_duplicates(subset='App', keep='first')

top10_apps = df.head(10)

print("Top 10 aplicativos por número de reviews:")
print(top10_apps[['App', 'Reviews']])
```

7 - Lista com 2 cálculos (top 10 apps com mais reviews e o app mais caro)

```python
import pandas as pd

df = pd.read_csv('googleplay_sem_duplicatas.csv')

df['Reviews'] = pd.to_numeric(df['Reviews'], errors='coerce')

df = df.dropna(subset=['Reviews'])

df = df.sort_values(by='Reviews', ascending=False).drop_duplicates(subset='App', keep='first')

top10_apps = df.head(10)

df['Price'] = pd.to_numeric(df['Price'].str.replace('$', '').str.replace('Free', '0'), errors='coerce')

df = df.dropna(subset=['Price'])

app_mais_caro = df.loc[df['Price'].idxmax()]

if app_mais_caro['App'] not in top10_apps['App'].values:
    top10_apps = pd.concat([top10_apps, app_mais_caro.to_frame().T], ignore_index=True)

print("Top 10 aplicativos por número de reviews e o aplicativo mais caro:")
print(top10_apps[['App', 'Reviews', 'Price']])
```

8 - Gráficos de barras do top 10 apps mais caros e gráfico de pizza das ratings dos apps no dataset

```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv('googleplay_sem_duplicatas.csv')

data['Price'] = data['Price'].apply(lambda x: float(x.replace('$', '')) if '$' in x else 0)

top10_expensive_apps = data.sort_values('Price', ascending=False).head(10)

plt.figure(figsize=(10, 5))
bars = plt.barh(top10_expensive_apps['App'], top10_expensive_apps['Price'], color='skyblue')
plt.xlabel('Preço em $')
plt.title('Top 10 Aplicativos Mais Caros')
plt.gca().invert_yaxis()  
for bar in bars:
    plt.text(bar.get_width(), bar.get_y() + bar.get_height()/2, f'${bar.get_width():.2f}', va='center')

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv('googleplay_sem_duplicatas.csv')

rating_counts = data['Rating'].value_counts()

plt.figure(figsize=(20, 12))
wedges, texts, autotexts = plt.pie(rating_counts, labels=rating_counts.index, autopct='%1.1f%%', startangle=140, colors=plt.cm.Paired(range(len(rating_counts))))

plt.legend(wedges, rating_counts.index, title="Ratings", loc="center left", bbox_to_anchor=(1, 0, 0.5, 1))

plt.title('Distribuição das Avaliações dos Aplicativos')
plt.axis('equal')  
plt.show()
```
